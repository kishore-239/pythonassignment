### 3. Write a function that takes two sets and returns their symmetric difference.
X = {1, 2, 3, 4}  
Y = {3, 4, 5, 6}
Hint: Use symmetric_difference() or ^.


```python
def symmetric_diff(set1, set2):
    return set1.symmetric_difference(set2)

X = {1, 2, 3, 4}
Y = {3, 4, 5, 6}

result = symmetric_diff(X, Y)
print(result)
```

### 6. Given three sets A, B, and C, find elements common to all three.
A = {1, 2, 3, 4, 5}  
B = {3, 4, 5, 6, 7}  
C = {0, 2, 3, 4, 8}
Hint: Use A & B & C or intersection().

```python
A = {1, 2, 3, 4, 5}
B = {3, 4, 5, 6, 7}
C = {0, 2, 3, 4, 8}

common_elements = A & B & C
print("Common elements (using &):", common_elements)

common_elements = A.intersection(B, C)
print("Common elements (using intersection()):", common_elements)
```

### 9. Given a set of numbers, write a function to remove all even numbers.
nums = {2, 3, 4, 5, 6, 7, 8, 9}
Hint: Use a loop with discard()

```python
nums = {2, 3, 4, 5, 6, 7, 8, 9}


def remove_evens(nums):
    for num in list(nums):
        if num % 2 == 0:
            nums.discard(num)
    return nums


print(remove_evens(nums))
```

### 11. Check if A is a subset of B and B is a superset of A.

### 13. From a list of integers, extract and print all unique digits using sets.
numbers = [123, 456, 789, 112]
Hint: Convert numbers to strings, extract digits.

```python
numbers = [123, 456, 789, 112]

digits = set()
for num in numbers:
    digits.update(str(num))

unique_digits = {int(d) for d in digits}

print("Unique digits:", unique_digits)
```

### 14. Copy a set and verify both have the same values but are different objects.
original_set = {'red', 'blue', 'green'}


```python
original_set = {'red', 'blue', 'green'}
copied_set = original_set.copy()

# Check value equality
if original_set == copied_set:
    print("The sets have the same values.")
else:
    print("The sets have different values.")

# Check object identity
if original_set is copied_set:
    print("They are the same object in memory.")
else:
    print("They are different objects in memory.")
```
