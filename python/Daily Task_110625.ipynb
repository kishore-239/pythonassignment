### 18. What happens if .find() doesn’t find the character?




```python
var = "Hello, world!"

char_to_find = 'z'

var = var.find(char_to_find)
```

```python
if var == -1:
    print(f"The character '{char_to_find}' was not found in the string.\n-If the character is not found, it returns -1.")
else:
    print(f"The character '{char_to_find}' was found at index {var}.")
```

#### Example

```python
k = "Hello, world!"
```

```python
print(k.find('z'))
```
