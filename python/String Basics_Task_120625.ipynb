### 94. How to replace only the second occurrence of 'a' in 'banana'?




```python
s = 'banana'
target = 'a'
replacement = 'X'
count = 2

indices = [i for i, char in enumerate(s) if char == target]

if len(indices) >= count:
    pos = indices[count-1]
    result = s[:pos] + replacement + s[pos+1:]
else:
    result = s

print(result)
```

### 96. Use .translate() to remove all vowels from 'interview'.




```python
s = 'interview'
vowels = 'aeiouAEIOU'

# Create a translation table mapping vowels to None
translation_table = str.maketrans('', '', vowels)
result = s.translate(translation_table)

print(result)  # Output: 'ntrvw'
```
