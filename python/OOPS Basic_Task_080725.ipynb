### 3. Write a class Circle that accepts radius as input and calculates area using a method get_area().


```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def get_area(self):
        return 3.14 * (self.radius ** 2)
```

```python
circle = Circle(1)
area = circle.get_area()
print(f"Area of the circle with radius {circle.radius} is: {area:.3f}")
```

### 5. Write a class Calculator with two attributes a and b. Add methods for add(), subtract(), multiply() and divide().


```python
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def multiply(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "Cannot divide by zero!"
        return self.a / self.b
```

```python
calc = Calculator(10, 5)

print(calc.add())
print(calc.subtract())
print(calc.multiply())
print(calc.divide())
```

### 6. Create a class Temperature that converts Celsius to Fahrenheit. Include a method to_fahrenheit()


```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9/5) + 32
```

```python
temp = Temperature(25)
fahrenheit = temp.to_fahrenheit()

print(fahrenheit)
```

### 8. Write a Python class Rectangle that calculates perimeter and area with separate methods.


```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def calculate_perimeter(self):
        return 2 * (self.length + self.width)
    
    def calculate_area(self):
        return self.length * self.width
```

```python
rect = Rectangle(5, 3)
print(f"Rectangle with length {rect.length} and width {rect.width}:")
print(rect.calculate_perimeter())
print(rect.calculate_area())
```

### 10. Explain the use of self in a class with an example.


#### self is a reference to the current instance of a class and is used to access its attributes and methods. 
#### It must be explicitly declared as the first parameter in instance methods

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")

    def birthday(self):
        self.age += 1
        print(f"Now I'm {self.age}.")
```

```python
kris = Person("kris", 22)
kish = Person("kish", 23)

kris.introduce()
kish.introduce()

kris.birthday()
```

### 11. What is the purpose of a constructor? Demonstrate with an example class Laptop.


#### The purpose of a constructor in a class is to:
###### 1️⃣ Initialize new objects with default or provided values
###### 2️⃣ Set up initial state by assigning values to attributes
###### 3️⃣ Validate inputs before object creation
###### 4️⃣ Perform setup operations (e.g., opening connections)

```python
class Laptop:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display(self):
        print(f"{self.brand} {self.model}")
```

```python
my_laptop = Laptop("samsung", "galaxy")  
my_laptop.display()
```
