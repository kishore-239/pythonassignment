#### Q1: Create a NumPy array with numbers from 1 to 20

#### Q2: Generate an array of 15 random integers between 10 and 50
- a) Find the mean, median, and standard deviation
- b) Sort the array in ascending order
- c) Replace all values greater than 30 with 30


```python
np.random.seed(9)
a = np.random.randint(10, 51, 15)
print(a)
```

```python
print(np.mean(a))
print(np.median(a))
print(np.std(a))
```

```python
b = np.sort(a)
print(b)
```

```python
c = np.where(a > 30, 30, a)
print(c)
```

#### Q5: Create a NumPy array with 100 values from 0 to 1 using linspace
- a) Find the sum of all values less than 0.5
- b) Count how many values are greater than 0.8


```python
a = np.linspace(0, 1, 100)
```

```python
b = np.sum(a[a < 0.5])
c = np.sum(a > 0.8)

print(b)
print(c)
```

#### Q8: Create two Series of integers with overlapping values
- a) Find the union, intersection, and difference
- b) Compare both Series element-wise for equality


```python
a = pd.Series([1, 2, 3, 4, 5])
b = pd.Series([4, 5, 6, 7, 8])

a
b
```

```python
u = pd.Series(np.union1d(a, b))
i = pd.Series(np.intersect1d(a, b))
d = pd.Series(np.setdiff1d(a, b))

print(u)
print(i)
print(d)
```

```python
c = a == b
print(c)
```

### Bonus: Conceptual Questions

#### Q11: What is the difference between a NumPy array and a Pandas Series?
- NumPy arrays are homogeneous and optimized for numerical operations.
- Pandas Series are built on top of NumPy arrays but allow labeled indexing and can hold heterogeneous data.

#### Q12: Can a Pandas Series hold different data types? Justify with examples.
- Yes, a Series can hold mixed types, but it will upcast to a common type (usually `object`).

```python
a_list = [1, "apple", 3.14, True]
print(a_list)
```

```python
a_pandas = pd.Series(a_list)
print(a_pandas)
```

#### Q13: How is broadcasting handled differently in NumPy vs Pandas?
- NumPy uses shape-based broadcasting.

- Pandas uses label-based alignment during operations, which can result in NaNs if labels don’t match.

#### Q14: What happens if you add two Series with different indexes? Give an example.
- Missing indexes will result in NaN.

```python
x = pd.Series([9, 20], index=["a", "b"])
y = pd.Series([18, 44], index=["b", "c"])
print(x + y)
```
