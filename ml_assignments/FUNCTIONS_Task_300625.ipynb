### 4. Define a function max_of_three(a, b, c) that returns the largest of the three numbers.




```python
def max_of_three(a, b, c):
    if a>=b and a>=c:
        return f"{a} is largest of the three numbers"
    elif b>=a and b>=c:
        return f"{b} is largest of the three numbers"
    else:
        return f"{c} is largest of the three numbers"
        
```

```python
max_of_three(3,6,9)
```

### 7. Define a function count_vowels(text) that returns the number of vowels in a string.




```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count
```

```python
count_vowels("kishore")
```

### 9. Create a function multiply_list(lst) that returns the product of all elements in a list.


```python
def multiply_list(lst):
    product = 1
    for num in lst:
        product *= num
    return product
```

```python
multiply_list([1,2,3,4,5])
```

### 11. Define a function is_palindrome(word) that checks whether a string is a palindrome.


```python
def is_palindrome(word):
    return word == word[::-1]
```

```python
is_palindrome("kisho")
```

```python
is_palindrome("madam")
```

### 12. Create a function word_frequency(text) that returns a dictionary with word counts in a given sentence.


```python
def word_frequency(text):
    word_list = text.lower().split()
    frequency = {}
    for word in word_list:
        word = word.strip(".,!?;:\"'()")
        if word:
            frequency[word] = frequency.get(word, 0) + 1
    return frequency
```

```python
print(word_frequency("Hello hello world"))

print(word_frequency("This is a test. Test this!"))
```

### 13. Write a function get_even_numbers(lst) that returns a list of even numbers from the given list.


```python
def get_even_numbers(lst):
    return [num for num in lst if num % 2 == 0]
```

```python
print(get_even_numbers([1, 2, 3, 4, 5, 6]))
```
