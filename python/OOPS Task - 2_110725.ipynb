## Inheritance

### 1. Create a class Person with attributes name and age. Create a subclass Student that adds student_id and displays all the information.




```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self,name,age,student_id):
        super().__init__(name,age)
        self.student_id = student_id

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Student Id: {self.student_id}")
```

```python
student = Student("kisho",22,"S2025")
student.display()
```

## 3. Multilevel Inheritance
 ### Create a class Grandparent, then subclass Parent, and then Child. Each should have a method show_generation(). Call them from the Child class object.




```python
class Grandparent:
    def show_generation(self):
        print("Hello my next two generations")

class Parent(Grandparent):
    def show_generation(self):
        print("Hello Iam the Parent")
        super().show_generation()

class Child(Parent):
    def show_generation(self):
        print("Iam the child")
        super().show_generation()
```

```python
child = Child()
child.show_generation()
```

## 4. Hierarchical Inheritance
 ### Create a class Shape, then derive classes Circle and Square with area calculation methods. Take user input and print the area.




```python
class Shape:
    def calculate_area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def calculate_area(self):
        return self.side ** 2
```

```python
circle = Circle(10)
print(circle.calculate_area())
```

```python
square = Square(10)
print(square.calculate_area()) 
```

## 7. Bank Account Simulation
### Create a class BankAccount with private balance. Add methods deposit(amount) and withdraw(amount) with validation to prevent overdrawing.




```python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance
    
    def deposit(self, amount):
        self.__balance += max(amount, 0)
        return ("Deposit amount must be positive", 
                f"Deposited INR {amount:.2f}. New balance: INR {self.__balance:.2f}")[amount > 0]
    
    def withdraw(self, amount):
        valid_amount = min(max(amount, 0), self.__balance)
        self.__balance -= valid_amount
        messages = [
            "Withdrawal amount must be positive",
            "Insufficient funds",
            f"Withdrew INR {amount:.2f}. New balance: INR {self.__balance:.2f}"
        ]
        return messages[(amount > 0) + (amount <= self.__balance)]
    
    def get_balance(self):
        return f"Current balance: INR {self.__balance:.2f}"  
```

```python
account = BankAccount(190)  
print(account.deposit(44))  
print(account.withdraw(44)) 
print(account.withdraw(44))
print(account.deposit(44)) 
print(account.get_balance())  
```

## 9. Update Marks Securely
###  Create a class Student with private __marks. Write a method to update marks only if the value is within 0–100 range.




```python
class Student:
    def __init__(self, name, initial_marks=0):
        self.name = name
        self.__marks = self.__validate_marks(initial_marks)
    
    def __validate_marks(self, marks):
        return max(0, min(100, marks))
    
    def update_marks(self, new_marks):
        self.__marks = self.__validate_marks(new_marks)
        return f"Marks updated to {self.__marks}"
    
    def get_marks(self):
        return self.__marks
```

```python

student = Student("kisho", 85)

print(student.update_marks(93))   
print(student.update_marks(150))  
print(student.update_marks(-20))  
print(student.get_marks())        
```

## 13. Function Overloading using Default Arguments
###  Create a class Calculator with a method add() that can add two or three numbers using default arguments.




```python
class Calculator:
    def add(self, a, b, c=None):
        if c is not None:
            return a + b + c
        return a + b
```

```python
calc = Calculator()
```

```python
print(calc.add(5, 3))    
print(calc.add(2, 4, 6)) 
```

## 14. Polymorphism with Inheritance
### Create a base class Operation with a method execute(). Create subclasses Add, Subtract, and Multiply that override execute().




```python
class Operation:
    def execute(self, a, b):
        pass

class Add(Operation):
    def execute(self, a, b):
        return a + b

class Subtract(Operation):
    def execute(self, a, b):
        return a - b

class Multiply(Operation):
    def execute(self, a, b):
        return a * b
```

```python
operations = [Add(), Subtract(), Multiply()]
```

```python
a, b = 18, 9
```

```python
for op in operations:
    print(f"{op.__class__.__name__}: {op.execute(a, b)}")
```

## 15. Operator Overloading
### Create a class Vector that has x and y coordinates. Overload the + operator to add two vector objects and return a new vector.


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
v1 = Vector(3,9)
v2 = Vector(15,21)
```

```python
v3 = v1 + v2
print(v3)
```
