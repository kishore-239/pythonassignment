### 8. What is the purpose of try, except, else, and finally blocks in Python?




##### These blocks are used for exception handling to manage errors gracefully and ensure clean resource management.

##### 1. try Block

##### Purpose: Contains code that might raise an exception.

```python
try:
    result = 10 / 0
except (ZeroDivisionError,) as e :
    print(e)
    print("Cannot divide by zero!")
```

##### 2. except Block

##### Purpose: Catches and handles specific exceptions.



```python
try:
    num = int(input("Enter a number: "))
except (ValueError, ) as e:
    print(e)
    print("Invalid input! Please enter a number.")
```

```python
try:
    file = open(r"C:\Users\kisho\OneDrive\Desktop\Task_180725\qsn_8.txt", "r")
    data = file.read()
except FileNotFoundError:
    print("File not found!")
except PermissionError:
    print("No permission to read the file!")
```

##### 3. else Block

##### Purpose: Runs only if no exceptions occur in try.



```python
try:
    age = int(input("Enter your age: "))
except ValueError:
    print("Invalid age!")
else:
    print(f"Your age is: {age}")
```

##### 4. finally Block


##### Purpose: Always executes, whether an exception occurs or not.


```python
try:
    file = open(r"C:\Users\kisho\OneDrive\Desktop\Task_180725\qsn_8.txt", "r")
    print(file.read())
except FileNotFoundError:
    print("File not found!")
else:
    print("There is no error!")
finally:
    file.close()
    print("Execution complete.")
```

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero!")
else:
    print(f"Result: {result}, This is ELSE block and compiled because of no error")
finally:
    print("This is FINALLY block always runs.")
```
