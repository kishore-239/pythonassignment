### 17. Reverse a number using a loop.




```python
num = int(input("Enter a number: "))
reversed_num = 0
original_num = num

num = abs(num)

while num > 0:
    digit = num % 10
    reversed_num = reversed_num * 10 + digit
    num = num // 10


if original_num < 0:
    reversed_num *= -1

print(f"Reversed number: {reversed_num}")
```

### 20. Check whether a number is an Armstrong number.




```python
def is_armstrong_number(num):

    num_str = str(num)
    n = len(num_str)
    
    sum_of_powers = sum(int(digit) ** n for digit in num_str)
    
    return sum_of_powers == num

number = int(input("Enter a number: "))
if is_armstrong_number(number):
    print(f"{number} is an Armstrong number.")
else:
    print(f"{number} is not an Armstrong number.")
```

### 21. Count vowels and consonants in a given string.




```python
def count_vowels_and_consonants(input_string):
    vowels = "aeiouAEIOU"
    vowel_count = 0
    consonant_count = 0
    
    for char in input_string:
        if char.isalpha():
            if char in vowels:
                vowel_count += 1
            else:
                consonant_count += 1
    
    return vowel_count, consonant_count


input_str = input("Enter a string: ")
vowels, consonants = count_vowels_and_consonants(input_str)
print(f"Vowels: {vowels}, Consonants: {consonants}")
```

### 22. Reverse a string without using slicing.




```python
def reverse_string_loop(s):
    reversed_str = ""
    for i in range(len(s) - 1, -1, -1):
        reversed_str += s[i]
    return reversed_str

input_str = "hello"
print(reverse_string_loop(input_str))
```

### 23. Check if a string is a palindrome.




```python
import re

def is_palindrome(s):

    cleaned_s = re.sub(r'[^a-zA-Z0-9]', '', s).lower()

    return cleaned_s == cleaned_s[::-1]

input_str = input("Enter a string: ")
if is_palindrome(input_str):
    print(f"'{input_str}' is a palindrome.")
else:
    print(f"'{input_str}' is not a palindrome.")
```

### 24. Remove all duplicates from a string.




```python
def remove_duplicates_set(s):
    seen = set()
    result = []
    for char in s:
        if char not in seen:
            seen.add(char)
            result.append(char)
    return ''.join(result)

input_str = "hello world"
print(remove_duplicates_set(input_str))
```

### 25. Count the frequency of each character in a string.




```python
def count_characters(s):
    frequency = {}
    for char in s:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    return frequency

input_str = "hello world"
char_frequency = count_characters(input_str)
print(char_frequency)
```

### 36. Write a function to check if a number is prime.




```python
import math

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True

print(is_prime(5))  
print(is_prime(4))   
print(is_prime(29)) 
```

### 38. Write a recursive function to calculate factorial.




```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = 5
print(f"The factorial of {num} is: {factorial(num)}")
```

### 39. Define a function that accepts a list and returns the second-largest number.




```python
def second_largest(numbers):
    if len(numbers) < 2:
        return None
    
    max1 = max2 = float('-inf') 
    
    for num in numbers:
        if num > max1:
            max2 = max1
            max1 = num 
        elif num > max2 and num != max1:
            max2 = num 
    
    if max2 == float('-inf'):
        return None
    
    return max2

numbers = [10, 5, 20, 20, 8]
print("Second-largest number:", second_largest(numbers))
```
